**Q5. What's the maximum fuel efficiency of cars from Asia?**
- answer: $41.2$

```$
SELECT
    origin,
    MAX(fuel_efficiency_mpg)
FROM cars
GROUP BY origin
HAVING origin = 'Asia'
```

````python
df.groupby('origin')['fuel_efficiency_mpg'].max()['Asia']
````

**Q6. Median value of horsepower**
1. Find the median value of the `horsepower` column in the dataset.
2. Next, calculate the most frequent value of the same `horsepower` column.
3. Use the `fillna` method to fill the missing values in the `horsepower` column with the most frequent value from the previous step.
4. Now, calculate the median value of `horsepower` once again.

**Has it changed?**
- Yes, it increased
- Yes, it decreased $\checkmark$
- No

````python
median = df['horsepower'].median()
median
````

````python
most_frequent_value = df['horsepower'].value_counts().index[0]
most_frequent_value
````

````python
new_median = df['horsepower'].fillna(most_frequent_value).median()
new_median
````

````python
median - new_median
````
